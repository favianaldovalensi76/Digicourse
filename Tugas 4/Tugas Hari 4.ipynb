<h2>Soal 1: Comparison Operator</h2>

- Berikan contoh comparison dari tipe data string
- Berikan contoh comparison dari tipe data boolean
- Berikan contoh comparison dari tipe data integer

```python
# Comparison dari tipe data string
string1 = "apple"
string2 = "banana"
print(string1 == string2)  # False
print(string1 < string2)   # True (berdasarkan urutan alfabet)

# Comparison dari tipe data boolean
bool1 = True
bool2 = False
print(bool1 == bool2)      # False
print(bool1 != bool2)      # True

# Comparison dari tipe data integer
num1 = 10
num2 = 20
print(num1 > num2)         # False
print(num1 <= num2)        # True
```

<h2>Soal 4: Comparison Operator dengan fungsi</h2>

Buatlah sebuah fungsi yang menerima satu argument bertipe data numeric dan menghasilkan sebuah return sebagai berikut :
- menghasilkan return 'Aneh' jika nilai dari argument tersebut adalah bilangan ganjil
- menghasilkan return 'Tidak Aneh' jika nilai dari argument tersebut adalah bilangan genap dan diantara nilai 2 sampai 5 (2 dan 5 termasuk)
- menghasilkan return 'Aneh' jika nilai dari argument tersebut adalah bilangan genap dan diantara nilai 6 sampai 20 (6 dan 20 termasuk)
- menghasilkan return 'Tidak Aneh' jika nilai dari argument tersebut adalah bilangan genap dan lebih besari dari 20

```python
# Buat fungsi
def cek_aneh(nilai):
    if nilai % 2 != 0:  # Jika bilangan ganjil
        return 'Aneh'
    elif nilai % 2 == 0 and 2 <= nilai <= 5:  # Jika bilangan genap di antara 2-5
        return 'Tidak Aneh'
    elif nilai % 2 == 0 and 6 <= nilai <= 20:  # Jika bilangan genap di antara 6-20
        return 'Aneh'
    elif nilai % 2 == 0 and nilai > 20:  # Jika bilangan genap lebih dari 20
        return 'Tidak Aneh'
    else:
        return 'Tidak valid'

# Contoh penggunaan fungsi
print(cek_aneh(3))   # Output: Aneh
print(cek_aneh(4))   # Output: Tidak Aneh
print(cek_aneh(18))  # Output: Aneh
print(cek_aneh(22))  # Output: Tidak Aneh
```

jawaban perbedaan

<h2>Soal 6: While Loop</h2>
    
Buatlah sebuah code while loop sebagai berikut :
- Buatlah suatu fungsi yang menerima satu input nilai numerik
- Dalam fungsi, Buat sebuah while-loop dengan melakukan comparasi terhadap inputan tersebut, dimana looping/iterasi terus berjalan apabila nilai inputan tidak sama dengan 0
- Dalam setiap iterasi update nilai variable inputan itu, jika input lebih dari 0, maka kurangi variabel tersebut sebesar 1, selain itu tambahkan 1
- print nilai variable input dalam setiap awal iterasi

```python
def fungsi_while(angka):
    while angka != 0:  # Loop akan berjalan selama angka tidak sama dengan 0
        print(angka)  # Cetak nilai angka
        if angka > 0:
            angka -= 1  # Kurangi angka sebesar 1 jika lebih dari 0
        else:
            angka += 1  # Tambahkan angka sebesar 1 jika kurang dari 0

# Contoh penggunaan
a = 10
print("Jika a = 10")
fungsi_while(a)

a = -10
print("\nJika a = -10")
fungsi_while(a)
```

<h2>Soal 7: For Loop</h2>

Lengkapi kode untuk menghasilkan suatu output yang diharapkan:





*   Cari siswa mana saja yang memiliki nilai lebih dari sama dengan 80. Masukkan kedalam sebuah list. print hasilnya






```python
data_siswa = [
    {
        "nama":"Budi",
        "nilai": 90
    },
    {
        "nama":"Nina",
        "nilai": 78
    },
    {
        "nama":"Rudi",
        "nilai": 91
    },
    {
        "nama":"Olivia",
        "nilai": 76
    },
    {
        "nama":"Leo",
        "nilai": 80
    },
    {
        "nama":"Liam",
        "nilai": 67
    },
    {
        "nama":"Sheila",
        "nilai": 76
    }
]
```

```python
#code here
data_siswa = [
    {"nama": "Budi", "nilai": 90},
    {"nama": "Nina", "nilai": 78},
    {"nama": "Rudi", "nilai": 91},
    {"nama": "Olivia", "nilai": 76},
    {"nama": "Leo", "nilai": 80},
    {"nama": "Liam", "nilai": 67},
    {"nama": "Sheila", "nilai": 76},
]

# Membuat list untuk siswa dengan nilai >= 80
siswa_lulus = []

# Menggunakan for loop untuk memeriksa setiap siswa
for siswa in data_siswa:
    if siswa["nilai"] >= 80:
        siswa_lulus.append(siswa["nama"])

# Cetak hasil
print(siswa_lulus)
```

Expected output:

['Budi', 'Rudi', 'Leo']
